```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def compute_summary_stats(values):
    """Utility functions for calculating summary metrics with robust checks."""
    if not values:
        return {'mean': 0.0, 'count': 0}
    return {
        'mean': sum(values) / len(values),
        'count': len(values)
    }
```

```python
# Generate updated experiment plot with cos overlay
x = np.linspace(0, 10, 100)
y = np.cos(x)
plt.plot(x, y, label='Cosine wave', color='orange')
plt.show()
```
